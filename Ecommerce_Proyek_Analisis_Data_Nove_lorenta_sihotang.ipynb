## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Kategori produk mana yang menghasilkan revenue tertinggi dan terendah selama tahun 2018?

  SMART-nya:

    - Specific: kategori produk + revenue
    - Measurable: total revenue
    - Action-oriented: dapat menjadi dasar untuk mengevaluasi performa kategori produk
    - Relevant: berkaitan langsung dengan performa penjualan
    - Time-bound: tahun 2018
- **Pertanyaan 2:** Seller mana yang memiliki jumlah pesanan terlambat terbanyak selama tahun 2018?

  SMART-nya:

  - Specific: seller + pesanan terlambat
  - Measurable: jumlah pesanan terlambat
  - Action-oriented: dapat digunakan untuk mengidentifikasi seller yang perlu dievaluasi dari sisi pemenuhan pesanan
  - Relevant: berkaitan dengan operasional e-commerce dan kualitas fulfillment
  - Time-bound: tahun 2018
- **Pertanyaan 3:** Kota mana yang memiliki jumlah order dan customer terbanyak selama tahun 2018?

  SMART-nya:

  - Specific: fokus pada kota, order, dan customer.
  - Measurable: jumlah order dan jumlah customer.
  - Action-oriented: dapat membantu mengidentifikasi wilayah dengan aktivitas - customer dan transaksi tertinggi.
  - Relevant: relevan untuk memahami distribusi pasar berdasarkan lokasi.
  - Time-bound: tahun 2018

### Cleaning Data

#### Fixing ... problem

```python
# Menangani missing value pada products_df pd product_category_name, product_name_lenght, product_description_lenght dst
# products_df.isna().sum()
products_df[products_df.product_category_name.isna()]
```

```python
products_df.product_category_name.value_counts()
# products_df.isna().sum()
```

```python
# menggunakan metode imputation untuk menangani missing value pada products
# Untuk product_category_name diisi dengan nilai string 'unknown'
# Untuk product_name_lenght, product_description_lenght, dan product_photos_qty: Isi dengan nilai 0 (menandakan tidak ada deskripsi/foto)
# Imputasi kategori dan metrik teks
products_df['product_category_name'] = products_df['product_category_name'].fillna('unknown')
products_df['product_name_lenght'] = products_df['product_name_lenght'].fillna(0)
products_df['product_description_lenght'] = products_df['product_description_lenght'].fillna(0)
products_df['product_photos_qty'] = products_df['product_photos_qty'].fillna(0)
```

```python
# karena product_weight_g, product_length_cm, product_height_cm, product_width_cm hanya hilang di 2 baris
# Imputasi dengan Median  (ga pengen dihapus)
cols = ['product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']
products_df[cols] = products_df[cols].fillna(products_df[cols].median())
```

```python
# products_df.describe()
products_df.isna().sum()
```

```python
# menangani masalah inacurrate pada tabel products_df kolom product_weight_g=0
# karena product yg weight yah = 0.0 hanya 4 (category:cama_mesa_banho) maka lebih baik dihapus saja dan lebih praktis
products_df = products_df[products_df['product_weight_g'] > 0]
```

```python
# menghitung revenue
# memfilter data yakni order status = delivered  dan tahun 2018
filter_date = orders_product_item_df[
    (orders_product_item_df['order_status'] == 'delivered') &
    (orders_product_item_df['order_purchase_timestamp'].dt.year.isin([2018]))
    ]
# menghitung revenue
category_name_revenue = (
    filter_date
    .groupby('product_category_name')['price']
    .sum()
    .reset_index()
    .sort_values(by='price', ascending=False)
)
# 4. Ambil kategori tertinggi dan terendah
top_category = category_name_revenue.iloc[0]
bottom_category = category_name_revenue.iloc[-1]

print("=== Kategori Revenue Tertinggi (2018) ===")
print(f"Kategori : {top_category['product_category_name']}")
print(f"Revenue  : {top_category['price']:,.2f}")

print("\n=== Kategori Revenue Terendah (2018) ===")
print(f"Kategori : {bottom_category['product_category_name']}")
print(f"Revenue  : {bottom_category['price']:,.2f}")
```

```python
sellers_order_item_df['ship_duration_days'] = (
    sellers_order_item_df['shipping_limit_date'] - sellers_order_item_df['order_purchase_timestamp']
).dt.total_seconds()/(24*3600)

sellers_order_item_df.groupby(by="seller_city").agg({
    "order_id": "nunique",
    "seller_id": "nunique",
    "order_delivered_carrier_date" :"max",
    "shipping_limit_date" :"max",
    "ship_duration_days": "mean",
    "price": "sum"
}).sort_values(by="price", ascending=False)
```

```python
# merge tbel orders dan customers
order_customers_df = pd.merge(
    left=orders_df,
    right= customers_df,
    how="left",
    left_on="customer_id",
    right_on="customer_id"
)
order_customers_df.head()
```

## Visualization & Explanatory Analysis

### Pertanyaan 1:

```python
# status =delivered & tahun 2018
df_2018 = all_table_df[
    (all_table_df['order_status'] == 'delivered') &
    (all_table_df['order_purchase_timestamp'].dt.year == 2018)
    ].copy()
# hapus baris kategorinya kosong
df_2018_clean = df_2018.dropna(subset=['product_category_name'])
# agregasi total revenue per produk
category_revenue_df = (
    df_2018_clean
    .groupby('product_category_name')
    .agg(
        total_orders=('order_id', 'nunique'),
        total_items_sold=('order_item_id', 'count'),
        total_revenue=('price', 'sum')
    )
    .reset_index()
    .sort_values(by='total_revenue', ascending=False)
)
print("TOP 5 kategori produk yang revenue tertinggi 2018")
display(category_revenue_df.head(5))
print("TOP 5 kategori produk yang revenue terendah 2018")
display(category_revenue_df.tail(5))
```

```python
# Set style
sns.set_theme(style="whitegrid")

# Figure dengan 2 subplot
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(16, 6))

#  top 5
top_5 = category_revenue_df.head(5)
sns.barplot(
    data=top_5,
    x='total_revenue',
    y='product_category_name',
    hue='product_category_name',
    palette='Blues_r',
    legend=False,
    ax=axes[0]
)
axes[0].set_title('Top 5 categories by revenue (2018)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Total revenue BRL')
axes[0].set_ylabel('Product category')

# Label angka di ujung bar Top 5
for p in axes[0].patches:
    width = p.get_width()
    axes[0].annotate(
        f'R${width/1000:,.1f}k',
        (width, p.get_y() + p.get_height() / 2.),
        ha='left', va='center',
        xytext=(8, 0),
        textcoords='offset points',
        fontsize=10
    )

# bottom 5
bottom_5 = category_revenue_df.tail(5).sort_values(by='total_revenue', ascending=True)
sns.barplot(
    data=bottom_5,
    x='total_revenue',
    y='product_category_name',
    hue='product_category_name',
    palette='Reds_r',
    legend=False,
    ax=axes[1]
)
axes[1].set_title('Bottom 5 categories by revenue 2018', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Total revenue BRL')
axes[1].set_ylabel('')

# Label angka di ujung bar Bottom 5
for p in axes[1].patches:
    width = p.get_width()
    axes[1].annotate(
        f'R$ {width:,.2f}',
        (width, p.get_y() + p.get_height() / 2.),
        ha='left', va='center',
        xytext=(8, 0),
        textcoords='offset points',
        fontsize=10
    )

plt.suptitle('E-Commerce Revenue Analysis by Product Category (2018)', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

### Pertanyaan 2:

```python
# no 2 seller mana yg memiliki jumlah pesanan terlambat terbanyak selama tahun 2018
df_2018 = all_table_df[
    (all_table_df['order_status'] == 'delivered') &
    (all_table_df['order_purchase_timestamp'].dt.year == 2018)].copy()
# filter order yg terlambat dikirim oleh seller
df_late_sellers = df_2018[df_2018['order_delivered_carrier_date'] > df_2018['shipping_limit_date']]
# agregasi jumlah order terlambat satu per ssatu
top_late_sellers_df = (
    df_late_sellers
    .groupby('seller_id')
    .agg(
        total_late_orders=('order_id', 'nunique'),
        total_late_items=('order_item_id', 'count')
    )
    .reset_index()
    .sort_values(by='total_late_orders', ascending=False)
)
# 10 seller paling lambat
print("TOP 10 seller dengan pesanan terlambat terbanyak 2018")
display(top_late_sellers_df.head(10))
```

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12,6))
# ambil 10 top
top_10_late =top_late_sellers_df.head(10)
# potong id seller jdi 8 karakter
top_10_late['seller_short_id'] = top_10_late['seller_id'].str[:8] + '...'
ax = sns.barplot(
    data=top_10_late,
    x='total_late_orders',
    y='seller_short_id',
    hue='seller_short_id',
    palette ='Reds_r',
    legend=False)
plt.title('Top 10 Sellers with Most Delayed Carrier Deliveries (2018)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Total Delayed Orders', fontsize=12)
plt.ylabel('Seller ID', fontsize=12)

for p in ax.patches:
    width = p.get_width()
    if width > 0:
        ax.annotate(
            f'{int(width)} orders',
            (width, p.get_y() + p.get_height() / 2.),
            ha='left', va='center',
            xytext=(8, 0),
            textcoords='offset points',
            fontsize=10
        )

plt.tight_layout()
plt.show()
```

### Pertanyaan 3:

```python
# no 3 kota mana yg memiliki jumlah order dan customer terbanyak selama tahun 2018
df_2018 = all_table_df[
    (all_table_df['order_status'] == 'delivered') &
    (all_table_df['order_purchase_timestamp'].dt.year ==2018)].copy()
# agregasi jumlah order dan custumer unik per kota
top_cities_df = (
    df_2018
    .groupby('customer_city')
    .agg(
        total_orders=('order_id','nunique'),
        total_customers=('customer_unique_id','nunique'))
    .reset_index().sort_values(by='total_orders', ascending=False))
print("top 10 kota dengan order dan customer terbanyak 2018")
display(top_cities_df.head(10))
```
